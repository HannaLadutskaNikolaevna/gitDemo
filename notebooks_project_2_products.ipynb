```python
import json

with open("data.json", "r", encoding="utf-8") as file:
    data = json.load(file)

print("Файл прочитан!")
```

```python
products = data["products"]
print("Количество товаров:", len(products))
print("Первый товар из списка:")
print(products[0])
```

```python
for item in products:
    title = item["title"]
    price = item["price"]
    print(f"Товар: {title} | Цена: ${price}")
```

```python
total_sum = 0
for item in products:
    total_sum += item["price"]
average_price = total_sum / len(products)

print(f"Общая стоимость всех товаров: ${round(total_sum, 2)}")
print(f"Средняя цена одного товара: ${round(average_price, 2)}")
```

```python
def analyze_products(products_list):
    print("=== АНАЛИЗ КАТАЛОГА ===")
    
    # 1. Считаем количество
    count = len(products_list)
    print(f"Всего товаров в каталоге: {count}")
    
    # 2. Считаем общую сумму
    total = 0
    for item in products_list:
        total += item["price"]
        
    # 3. Находим среднее
    avg = total / count
    
    print(f"Общая стоимость: ${round(total, 2)}")
    print(f"Средняя цена: ${round(avg, 2)}")

# Вызываем функцию
analyze_products(products)
```
